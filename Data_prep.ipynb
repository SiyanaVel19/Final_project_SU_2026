# Data 2: The central Banks policy rates

The central Banks policy rates -  set tracks the evolution of policy rates across the world. The policy rate is an important instrument used by central banks to implement their monetary policy. The data DOES NOT ACCOUNT for inflation, so we will need to get the data for inflation next :$Real\ Rate = Nominal\ Policy\ Rate - Inflation\ Rate$


```python
CBPR_raw = pd.read_csv("raw_data/Central_bank_policy_rates.csv", skiprows=5)
CBPR = CBPR_raw[['REF_AREA:Reference area', 'TIME_PERIOD:Period','OBS_VALUE:Value']].copy()
CBPR.columns = ['Country_Raw', 'Monthly', 'Nominal Interest Rate']
```

```python
CBPR
```

```python
CBPR['Country'] = CBPR['Country_Raw'].str.split(':').str[1]
CBPR = CBPR[['Country', 'Monthly', 'Nominal Interest Rate']].copy()
```

```python
CBPR
```

Because the data is monthly for the interest rates, but quarterly for the house price index, I have to make a mean for every quarter

1. Check if the monthly data is an 'object' or 'datatime'

```python
print(CBPR['Monthly'].dtype)
```

```python
CBPR['Monthly'] = pd.to_datetime(CBPR['Monthly'])
print(CBPR['Monthly'].dtype)
```

2. Now that is a datetime, I make the quarterly mean with the same quarters as the HPI table (Year-03-31, Year-6-30, Year-9-30 and Year-12-31)

 - Add a new column that indicates which month is in which Quater 

```python
CBPR.loc[:, 'temp_quater'] = CBPR['Monthly'].dt.to_period('Q')
```

```python
CBPR
```

```python
CBPR = CBPR.groupby(['Country', 'temp_quater'])['Nominal Interest Rate'].mean().reset_index()
CBPR = CBPR.copy()
CBPR.loc[:, 'Quarter'] = CBPR['temp_quater'].dt.to_timestamp('Q')
```

```python
CBPR
```

```python
final_columns = ['Country', 'Quarter', 'Nominal Interest Rate'] 
CBPR_clean = CBPR[final_columns]
```

```python
CBPR_clean
```

The country name bank for Interest Rates:

```python
CBPR_countries = set (CBPR_clean['Country'].unique())
print(f"Total countries in Central Bank Policy Rates: {len(CBPR_countries)}")
```

# Merge into a master table

To merge, I need all the quarter data be datetime

```python
HPI_clean['Quarter'] = pd.to_datetime(HPI_clean['Quarter'])
CPI_clean['Quarter'] = pd.to_datetime(CPI_clean['Quarter'])
CBPR_clean['Quarter'] = pd.to_datetime(CBPR_clean['Quarter'])
GDP_growth_clean['Quarter'] = pd.to_datetime(GDP_growth_clean['Quarter'])
Unemployment_clean['Quarter'] = pd.to_datetime(Unemployment_clean['Quarter'])
```

```python
master = pd.merge(HPI_clean, GDP_growth_clean, on=['Country', 'Quarter'], how='inner')

# Add Unemployment
master = pd.merge(master, Unemployment_clean, on=['Country', 'Quarter'], how='inner')

# Add Central Bank Policy Rates (CBPR)
master = pd.merge(master, CBPR_clean, on=['Country', 'Quarter'], how='inner')

# Add CPI
master = pd.merge(master, CPI_clean, on=['Country', 'Quarter'], how='inner')

# Sort to keep everything organized
master = master.sort_values(by=['Country', 'Quarter']).reset_index(drop=True)
```

```python
master
```

# Finding the 'real' interest rates & HPI growth

###  HPI growth for Year-to-year change

Right now, we have data about the 'level' of the Real Price Index, but we need to know how much it grows quarter to quarter to compare it to the real interest and inflation rate

We group countries so the calculations don't leak from the last entry of one country to another

```python
master['HPI_Growth'] = master.groupby('Country')['Real_HPI'].pct_change(periods=4) * 100
```

### Calculating the Real Interest Rate (Fisher Eq.)

The data we currently have for interest rates does not account for inflation, but we just got the quarterly inflation change
We will get the real interest rate by : $Real\ Rate = Nominal\ Policy\ Rate - Inflation\ Rate$


(Nominal - does not account for inflation; Real - accounts for inflation)

```python
master['Real_Rate'] = master['Nominal Interest Rate'] - master['Change of inflation rate']
```

```python
master = master.dropna().reset_index(drop=True)
```
