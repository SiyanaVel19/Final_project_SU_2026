# Data 2: The central Banks policy rates

The central Banks policy rates -  set tracks the evolution of policy rates across the world. The policy rate is an important instrument used by central banks to implement their monetary policy. The data DOES NOT ACCOUNT for inflation, so we will need to get the data for inflation next :$Real\ Rate = Nominal\ Policy\ Rate - Inflation\ Rate$


```python
CBPR_raw = pd.read_csv("raw_data/Central_bank_policy_rates.csv", skiprows=5)
CBPR = CBPR_raw[['REF_AREA:Reference area', 'TIME_PERIOD:Period','OBS_VALUE:Value']].copy()
CBPR.columns = ['Country_Raw', 'Monthly', 'Interest Rate']
```

```python
CBPR
```

```python
CBPR['Country'] = CBPR['Country_Raw'].str.split(':').str[1]
CBPR = CBPR[['Country', 'Monthly', 'Interest Rate']].copy()
```

```python
CBPR
```

Because the data is monthly for the interest rates, but quarterly for the house price index, I have to make a mean for every quarter

1. Check if the monthly data is an 'object' or 'datatime'

```python
print(CBPR['Monthly'].dtype)
```

```python
CBPR['Monthly'] = pd.to_datetime(CBPR['Monthly'])
print(CBPR['Monthly'].dtype)
```

2. Now that is a datetime, I make the quarterly mean with the same quarters as the HPI table (Year-03-31, Year-6-30, Year-9-30 and Year-12-31)

 - Add a new column that indicates which month is in which Quater 

```python
CBPR.loc[:, 'temp_quater'] = CBPR['Monthly'].dt.to_period('Q')
```

```python
CBPR
```

```python
CBPR = CBPR.groupby(['Country', 'temp_quater'])['Interest Rate'].mean().reset_index()
CBPR = CBPR.copy()
CBPR.loc[:, 'Quarter'] = CBPR['temp_quater'].dt.to_timestamp('Q')
```

```python
CBPR
```

```python
final_columns = ['Country', 'Quarter', 'Interest Rate'] 
CBPR_clean = CBPR[final_columns]
```

```python
CBPR_clean
```

# Data 3: Consumer prices


The consumer price index data set contains long series that track the overall change in the prices of goods and services purchased by households. Consumer price indices are a key macroeconomic indicator, for instance to measure inflation and assess purchasing power over time.

```python
CPI_raw = pd.read_csv("raw_data/Consumer_prices_Yearly_change.csv", skiprows=5)#CPI_raw = Consumer prices index raw
CPI = CPI_raw[['REF_AREA:Reference area', 'TIME_PERIOD:Period','OBS_VALUE:Value']].copy()
CPI.columns = ['Country_Raw', 'Monthly', 'Change of inflation rate']
```

```python
CPI
```

```python
CPI['Country'] = CPI['Country_Raw'].str.split(':').str[1]
CPI = CPI[['Country', 'Monthly', 'Change of inflation rate']].copy()
```

The data for the consumer prices is monthly, but I need it in quarters

```python
print(CPI['Monthly'].dtype)
```

2. It's an object so we change it to datetime

```python
CPI['Monthly'] = pd.to_datetime(CPI['Monthly'])
print(CPI['Monthly'].dtype)
```

4. Now that is a datetime, I make the quarterly mean with the same quarters as the HPI & CBPR table (Year-03-31, Year-6-30, Year-9-30 and Year-12-31)

I insert a new column that indicates if the month belongs to Q1, Q2, Q3 or Q4

```python
CPI.loc[:, 'temp_quater'] = CPI['Monthly'].dt.to_period('Q')
```

```python
CPI = CPI.groupby(['Country', 'temp_quater'])['Change of inflation rate'].mean().reset_index()
```

```python
CPI.loc[:, 'Quarter'] = CPI['temp_quater'].dt.to_timestamp('Q')
final_columns = ['Country', 'Quarter', 'Change of inflation rate'] 
CPI_final = CPI[final_columns]
```
